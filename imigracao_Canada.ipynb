```python
import pandas as pd
```

```python
df = pd.read_csv('/content/canadian_immegration_data.csv')
```

```python
df
```

```python
df.info()
```

```python
df.set_index('Country', inplace=True)
```

```python
anos = list(map(str, range(1980, 2014)))
```

```python
anos
```

```python
brasil = df.loc['Brazil', anos]
```

```python
brasil
```

```python
brasil_dict = {'ano': brasil.index.tolist(),'imigrantes': brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
dados_brasil
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigração do Brasil para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Número de Imigrantes')
plt.grid(True)
plt.xticks(['1980','1985','1990','1995','2000','2005','2010'])
plt.show()
```

```python
argentina = df.loc['Argentina', anos]
argentina_dict =  {'ano': argentina.index.tolist(),'imigrantes': argentina.values.tolist()}
dados_argentina = pd.DataFrame(argentina_dict)
dados_argentina
```

```python
plt.plot(dados_argentina['ano'], dados_argentina['imigrantes'])
plt.plot(dados_brasil['imigrantes'])
plt.xticks(['1980','1985','1990','1995','2000','2005','2010','2015'])
plt.legend(['Argentina', 'Brasil'])
plt.title('Imigração da Argentina e do Brasil para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Número de Imigrantes')
plt.grid(True)
plt.show()
```

```python
IPython_default = plt.rcParams.copy()
plt.style.use('fivethirtyeight')
```

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Imigração do Brasil para o Canadá \n1980 e 2013', fontsize=18, loc='left')
ax.set_xlabel('Ano', fontsize=14)
ax.set_ylabel('Número de Imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

ax.grid(True)

plt.show()
```

```python
fig, axs= plt.subplots(1,2, figsize=(25,10))
axs[0].plot(dados_brasil['ano'], dados_brasil['imigrantes'], color='r')
axs[0].set_title('Imigração do Brasil para o Canadá entre 1980 e 2013')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].set_xlabel('Ano')
axs[0].set_ylabel('Número de Imigrantes')
axs[0].grid(True)

axs[1].boxplot(dados_brasil['imigrantes'])
axs[1].set_title('Boxplot da Imigração do Brasil para o Canadá entre 1980 e 2013')
axs[1].set_xlabel('Brasil')
axs[1].set_ylabel('Número de Imigrantes')
axs[1].grid(True)

plt.show()
```

```python
fig, axs = plt.subplots(2,2, figsize=(15,10))
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle('Imigração dos 4 maiores países da América do Sul para o Canadá entre 1980 e 2013', fontsize=16)

axs[0,0].plot(df.loc['Brazil', anos])
axs[0,0].set_title('Brasil')

axs[0,1].plot(df.loc['Colombia', anos])
axs[0,1].set_title('Colombia')

axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Peru', anos])
axs[1,1].set_title('Peru')

for ax in axs.flat:
    ax.xaxis.set_major_locator(plt.MultipleLocator(5))

for ax in axs.flat:
    ax.set_xlabel('Ano')
    ax.set_ylabel('Número de Imigrantes')
    ax.grid(True)

ymin = 0
ymax = 7000

for ax in axs.ravel():
    ax.set_ylim(ymin, ymax)

plt.show()
```

```python
america_sul = df.query('Region=="South America"')
america_sul
```

```python
cores = []

for pais in america_sul_sorted.index:
  if pais=='Brazil':
    cores.append('g')
  else:
    cores.append('silver')

ig, ax = plt.subplots(figsize=(12,6))
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color= cores)
ax.set_title('Imigração dos países da América do Sul para o Canadá\n 1980 e 2013', loc='left', fontsize=18)
ax.set_xlabel('Número de Imigrantes', fontsize=14)
ax.set_ylabel('', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

for i, v in enumerate(america_sul_sorted['Total']):
  ax.text(v + 30, i, str(v), color='black', fontsize=10, ha='left', va= 'center')

ax.set_frame_on(False)
ax.get_xaxis().set_visible(False)

ig.savefig('imig_america_sul_canada.png', transparent=False, dpi=300, bbox_inches='tight')
plt.show()
```

```python
import seaborn as sns
```

```python
top_10 = df.sort_values('Total', ascending=False).head(10)
top_10
```

```python
sns.barplot(data=top_10, y=top_10.index, x='Total', orient='h')
plt.title('Top 10 países que mais migraram para o Canadá')
```

```python
ax= sns.barplot(data= top_10, y=top_10.index, x='Total', orient='h')
ax.set(title = 'Países com maior imigração para o Canadá \n 1980 a 2013',
      xlabel = 'Número de Imigrantes',
      ylabel = '')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax = sns.barplot(data=top_10, y=top_10.index, x='Total', orient='h')
ax.set_title('Países com maior imigração para o Canadá \n 1980 a 2013', loc='left', fontsize=16)
ax.set_xlabel('Número de Imigrantes', fontsize=14)
ax.set_ylabel('')
plt.show()
```

```python
def gerar_grafico_paleta(palette):
  fig, ax = plt.subplots(figsize=(10,5))
  ax = sns.barplot(data=top_10, y=top_10.index, x='Total', orient='h', palette=palette)
  ax.set_title('Países com maior imigração para o Canadá \n 1980 a 2013', fontsize=16, loc='left')
  ax.set_xlabel('Número de Imigrantes', fontsize=14)
  ax.set_ylabel('')
  plt.show()
```
